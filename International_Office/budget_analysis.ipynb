## Load the Data Set

```python
import matplotlib.pyplot as plt
import pandas as pd

International = pd.read_csv("International_Budget.csv")
International_bu_df = International.set_index("Academic Year")
International_money_stuff  = International_bu_df.loc[:, "Total Revenue": "Budget"]

enrollemnt = International_bu_df.loc[:, "Total number of nationalities": "Returning Students"]

budget_comparison = International_bu_df[["Budget"]]
student_com = International_bu_df[["Total number of nationalities", "New Students"]]
```

```python
fig, axes = plt.subplots(nrows = 2, ncols=2, figsize = (16,10))

axes[0, 0].grouped_bar(International_money_stuff, colors = ["#131247", "#3CD0E7", "#2ECC71", "#3498DB"])
axes[0, 0].legend(loc='upper left');

axes[0, 1].grouped_bar(enrollemnt, colors = ["#E67E22", "#9B59B6", "#E84393", "#795548"])
axes[0, 1].legend(loc='upper left');

axes[1, 0].grouped_bar(budget_comparison, colors = ["#3498DB"])
axes[1, 0].legend(loc='upper center');

axes[1, 1].grouped_bar(student_com, colors = ["#E67E22","#E84393"])
axes[1, 1].legend(loc='upper left');
```

```python
df = International_bu_df.copy()

if 'Academic Year' in df.columns:
    df = df.set_index('Academic Year')

money_cols = [
    'Total Revenue',
    'Total New Students Revenue',
    'Total Returning Students Revenue',
    'Budget'
]

count_cols = [
    'Total number of nationalities',
    'Total Number of Students',
    'New Students',
    'Returning Students'
]

formatters = {
    col: '${:,.2f}'.format for col in money_cols if col in df.columns
}

formatters.update({
    col: '{:,.0f}'.format for col in count_cols if col in df.columns
})

styled_table = (
    df.style
    .format(formatters)
    .set_table_styles([
        {
            'selector': 'caption',
            'props': [
                ('font-size', '20px'),
                ('font-weight', 'bold'),
                ('color', '#14532D'),
                ('padding', '16px')
            ]
        },
        {
            'selector': 'th',
            'props': [
                ('background-color', '#14532D'),
                ('color', 'white'),
                ('padding', '14px'),
                ('text-align', 'center'),
                ('border', '1px solid #14532D')
            ]
        },
        {
            'selector': 'td',
            'props': [
                ('padding', '12px'),
                ('text-align', 'center'),
                ('border', '1px solid #D1D5DB')
            ]
        },
        {
            'selector': 'tbody tr:nth-child(even)',
            'props': [
                ('background-color', '#DCFCE7')
            ]
        },
        {
            'selector': 'tbody tr:nth-child(odd)',
            'props': [
                ('background-color', '#F0FDF4')
            ]
        }
    ])
    .set_properties(**{
        'color': '#166534',
        'font-size': '13px'
    })
)

display(styled_table)
```
