## Load the Data Set

```python
import matplotlib.pyplot as plt
import pandas as pd

International = pd.read_csv("International_Budget.csv")
International_bu_df = International.set_index("Academic Year").loc[:, "Total Revenue": "Budget"]
International_required  = International_bu_df
International_required
```

```python
fig, axes = plt.subplots()
axes.grouped_bar(International_required)
axes.legend(loc='upper left');
```
